```python
import numpy as np
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import load_model
```

```python
#loading the dataset
word_index = imdb.get_word_index()
reversed_words = {value:key for key, value in word_index.items()}
```

```python
model = load_model('simple_rnn_imdb.h5')
model.summary()
```

```python
# helper functions

def decode_review(encoded_review):
    return ' '.join([reversed_words.get(i-3, '?') for i in encoded_review])

#function to preprocess the user input
def preprocess(text):
    words = text.lower().split() # lowers and splits the input sentence to lists
    encoded_review = [word_index.get(word, 2) + 3  for word in words] #tokenize the each words from the list by taking th the key words and returns the index from the imdb
    padded_review = sequence.pad_sequences([encoded_review], maxlen=500, padding='pre') # padds the encoded review by 500 words
    return padded_review
```

```python
# prediction function

def predict_sentiment(review):
    padded_input = preprocess(review)
    prediction = model.predict(padded_input)
    sentiment = 'Positive' if prediction[0][0] > 0.5 else 'Negetive'
    return sentiment, prediction[0][0]
```

```python
example_review = "i dont have words to describe"
sentiment, prediction = predict_sentiment(example_review)
print(f'Review: {example_review}')
print(f'Sentiment: {sentiment}')
print(f'Prediction score: {prediction}')
```
